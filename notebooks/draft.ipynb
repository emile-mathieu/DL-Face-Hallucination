We can use this file to draft and test code for our project. We can also use it to organize our thoughts and plan the structure of our codebase :)
Feel free to change anything guys, the structure etc.!
 
I thought to organize the code in the following way when training and testing the model on GPU CLUSTER:
1. `dataset.py` for data loading and preprocessing.
2. `models/model.py` file.
3. `train.py` and `test.py` for training and testing the model respectively.
4. The `main.py` will be used to start training and testing, main entry point for everything.
